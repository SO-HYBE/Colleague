```python
from sklearn.linear_model import LogisticRegression
from sklearn.multiclass import OneVsRestClassifier
from sklearn import datasets
from sklearn.preprocessing import StandardScaler
```

```python
iris = datasets.load_iris()
x = iris.data
y = iris.target
```

```python
sclaer = StandardScaler()
x_std = sclaer.fit_transform(x)
```

```python
clf = OneVsRestClassifier(LogisticRegression(random_state=0))
```

```python
model = clf.fit(x_std , y)
```

### 1- Explain Each in the logistic regression model:

Different Solvers params:

- lbfgs (Default): Approximates the second derivative matrix updates. Good for multiclass problems and most standard datasets.

- liblinear: Uses a coordinate descent algorithm. Good for small datasets and handling L1 penalties. Only supports One-vs-Rest for multiclass.

- newton-cg: Uses a Newton-Raphson method (calculates the exact Hessian matrix). Handles L2 or no penalty and supports multiclass.

- sag: Stochastic Average Gradient descent. Faster for large datasets (both features and samples).

- saga: A variant of sag that also supports the L1 penalty (useful for sparse multinomial logistic regression).

- newton-cholesky: An exact Newton solver that is a good choice for datasets with far more samples than features.

Different Multiclass params:

- ovr (One-vs-Rest): Fits a separate binary classifier for each class against all other classes combined.
 
- multinomial: Minimizes the multinomial loss fit across the entire probability distribution. It calculates the joint probability of all classes simultaneously.

- auto: Automatically selects ovr if the solver is liblinear or if the data is binary; otherwise, it selects multinomial.

### 2- Calculate the accuracy of the model

## 3- Make manual standard scaling funcation from the calculated standard scaler of sklearn:

#### A- Make a function that takes a scaler as input and returns a json file with the scaling values as keys and values.

#### B- Make another function that takes the json file and the raw input and returns the input values after scaling.

```python
import json

def export_scaler_to_json(scaler, filename="scaler_params.json"):
    """Extracts mean and scale from a fitted StandardScaler and saves to JSON."""
    scaling_params = {
        "mean": scaler.mean_.tolist(),
        "scale": scaler.scale_.tolist()
    }
    
    with open(filename, 'w') as f:
        json.dump(scaling_params, f, indent=4)
        
    return scaling_params

params = export_scaler_to_json(sclaer, "scaler_params.json")
print(params)
```

```python
import numpy as np

def manual_transform(json_filename, raw_input):
    """Loads scaling params from JSON and applies standard scaling to raw input."""
    with open(json_filename, 'r') as f:
        params = json.load(f)
        
    mean = np.array(params["mean"])
    scale = np.array(params["scale"])
    raw_input_array = np.array(raw_input)
    
    scaled_output = (raw_input_array - mean) / scale
    
    return scaled_output

scaled_n = manual_transform("scaler_params.json", n)
print(scaled_n)
print(model.predict(scaled_n))
```
