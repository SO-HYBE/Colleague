## Added train-test split for the data to get the accuracy of the test set and added the gini criterion to compare it with entropy.

```python
X_train = X_train.copy()
X_test = X_test.copy()

X_train['company_n'] = le_company.fit_transform(X_train['company'])
X_train['job_n'] = le_job.fit_transform(X_train['job'])
X_train['degree_n'] = le_degree.fit_transform(X_train['degree'])

X_test['company_n'] = le_company.transform(X_test['company'])
X_test['job_n'] = le_job.transform(X_test['job'])
X_test['degree_n'] = le_degree.transform(X_test['degree'])
```

In machine learning, Gini Impurity and Entropy are both metrics used to measure the "disorder" or "impurity" in a dataset. They help a Decision Tree decide which feature provides the best split at each node.

1. Mathematical Rules


Gini Impurity$$G = 1 - \sum_{i=1}^{n} (p_i)^2$$Entropy$$H = -\sum_{i=1}^{n} p_i \log_2(p_i)$$

2. Core Differences


Computational Complexity: Gini is faster to calculate because it only involves squaring probabilities. Entropy is more computationally expensive because it requires calculating a logarithm ($\log_2$). Sensitivity: Entropy is slightly more sensitive to changes in node probabilities than Gini. This means it often results in more "balanced" trees, whereas Gini tends to isolate the most frequent class into its own branch. Resulting Trees: In most practical cases, both metrics produce very similar trees. If you are working on a large-scale project where performance matters, Gini is usually the preferred default.
