```python
import tensorflow as tf
```

```python
from keras.datasets import mnist
from keras.models import Sequential
from keras.layers import Dense, Dropout
from keras import layers, models
from keras.initializers import RandomNormal
from keras.callbacks import EarlyStopping
from tensorflow.keras.preprocessing.image import ImageDataGenerator # type: ignore
from tensorflow.keras.callbacks import ReduceLROnPlateau # type: ignore
```

```python
(x_train , y_train) , (x_test,y_test) = mnist.load_data()
x_train.shape
```

```python
x_train = x_train.reshape(x_train.shape[0],28*28)
x_test = x_test.reshape(x_test.shape[0],28*28)
x_train.shape
```

```python
x_train = x_train / 255.0
x_test = x_test / 255.0
x_train.shape
```

```python
w_init = RandomNormal(mean=1,stddev=1)
```

# Adjusted Neural Network for better accuracy as per Assignment:


```python
model = Sequential()
model.add(Dense(128, activation= 'relu', kernel_initializer =w_init , input_dim=28*28))
model.add(Dense(256, activation= 'relu', kernel_initializer =w_init ))
model.add(layers.BatchNormalization())
model.add(Dropout(0.2))
model.add(Dense(10, activation= 'softmax', kernel_initializer =w_init))
model.compile(loss = 'sparse_categorical_crossentropy' , optimizer='AdamW' , metrics=['accuracy'])
model.summary()
```

```python
early_stop = EarlyStopping(
    monitor='val_loss', 
    patience=5, 
    restore_best_weights=True
)
```

```python
history = model.fit(x_train,y_train,epochs=100 , callbacks=[early_stop], validation_split=0.2, batch_size=16)
```

```python
loss, accuracy = model.evaluate(x_test, y_test)
print(accuracy)
```

#Task: Enhance the accuracy to an acceptable level with NN and CNN


### As shown, Vanilla Neural Networks are not the best for dealing with images, and the accuracies are not as expected. So, I will create another model optimized to the max to get the highest possible accuracy using Convolutional Neural Networks, Batch Norm, and Dropouts


```python
(x_train_cnn , y_train_cnn) , (x_test_cnn,y_test_cnn) = mnist.load_data()
```

```python
datagen = ImageDataGenerator(
    rotation_range=10,
    zoom_range=0.1,
    width_shift_range=0.1,
    height_shift_range=0.1
)
x_train_cnn = x_train_cnn.reshape(-1, 28, 28, 1)
x_test_cnn = x_test_cnn.reshape(-1, 28, 28, 1)

x_train_cnn = x_train_cnn.astype('float32') / 255.0
x_test_cnn = x_test_cnn.astype('float32') / 255.0

datagen.fit(x_train_cnn)
```

```python
learning_rate_reduction = ReduceLROnPlateau(monitor='val_accuracy', 
                                            patience=3, 
                                            verbose=1, 
                                            factor=0.5, 
                                            min_lr=0.00001)
```

```python
model = models.Sequential([
    # Block 1
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.BatchNormalization(),
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.25),

    # Block 2
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.BatchNormalization(),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.25),

    # Output Layer
    layers.Flatten(),
    layers.Dense(512, activation='relu'),
    layers.BatchNormalization(),
    layers.Dropout(0.5),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy']) 

model.summary()
```

```python
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

print(f"New shape for training: {x_train.shape}")
```

```python
history = model.fit(x_train,y_train,epochs=10 , callbacks=[early_stop,learning_rate_reduction], validation_split=0.2, batch_size=16)
```

```python
cnn_loss, cnn_accuracy = model.evaluate(x_test, y_test)
print(cnn_accuracy)
```
