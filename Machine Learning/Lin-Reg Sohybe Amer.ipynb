```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv("Task8-data.csv")
```

```python
data.head(10)
```

```python
data.info()
```

```python
data.describe()
```

```python
sns.pairplot(data,x_vars=["YearsExperience"],y_vars=["Salary"],size=5,
             kind="scatter")
plt.xlabel("years")
plt.ylabel("salary")
plt.title("salary prediction")
plt.show()
```

```python
# data.iloc[:,:-1]
X = data.iloc[:,:-1]
y = data.iloc[:,-1]
```

```python
X_train , X_test , y_train , y_test = train_test_split(X,y,train_size=0.8,random_state = 10)

my_model = LinearRegression()
my_model.fit(X_train,y_train)
```

```python
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("Predictions shape:", my_model.predict(X_train).shape)
```

```python
plt.scatter(X_train, y_train, color = 'red')
plt.plot(X_train, my_model.predict(X_train), color = 'blue')
plt.title('Salary vs Experience (Training set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

```python
from sklearn.metrics import mean_squared_error
y_pred = my_model.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error:", mse)
squareroot_mse = np.sqrt(mse)
print("Root Mean Squared Error:", squareroot_mse)
```
